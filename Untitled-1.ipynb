```python
import cv2
import os
import sys

# ====================== SETTINGS ======================
# Ensure this path is correct and the file exists
video_path = r"C:\Users\Admin\Downloads\projects\vehicle_plate_detection\IN_LANE_3_BACK_ZOOM_NVR_20260429104732_20260429114529_1999865.mp4"

input_folder = r"C:\Users\Admin\Downloads\projects\vehicle_plate_detection\data\frames"
max_frames_to_save = 5000  # STOP after this many frames
skip_seconds = 8          # Save 1 frame every 10 seconds
# =====================================================

# Check if file exists
if not os.path.exists(video_path):
    print(f"❌ ERROR: Video file not found at: {video_path}")
    sys.exit()

# Create folder if it doesn't exist
os.makedirs(input_folder, exist_ok=True)

# Open video
cap = cv2.VideoCapture(video_path)

# Check if camera opened successfully
if not cap.isOpened():
    print("❌ ERROR: Could not open video file. Try installing ffmpeg or converting the video.")
    print("Hint: Run 'pip install opencv-python-headless'")
    sys.exit()

fps = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if fps == 0 or total_frames == 0:
    print("❌ ERROR: Could not determine FPS or Frame Count. The video may be corrupt or use an unsupported codec.")
    print("Try converting the video to H.264 MP4 using VLC or Handbrake.")
    cap.release()
    sys.exit()

print(f"✅ Video Loaded Successfully!")
print(f"   FPS: {fps:.2f}")
print(f"   Total Frames: {total_frames}")
print(f"   Duration: {total_frames/fps:.2f} seconds")
print(f"   Saving 1 frame every {skip_seconds} second(s)...")
print(f"   Stopping after {max_frames_to_save} frames.\n")

frame_count = 0
saved_count = 0
last_saved_time = -skip_seconds

while True:
    ret, frame = cap.read()
    if not ret:
        print("End of video reached.")
        break

    frame_count += 1
    current_time_sec = frame_count / fps

    # Check if we have reached the limit
    if saved_count >= max_frames_to_save:
        print(f"\n Limit reached: {max_frames_to_save} frames saved.")
        break

    # Save frame based on time interval
    if current_time_sec >= last_saved_time + skip_seconds:
        
        last_saved_time = current_time_sec
        saved_count += 1
        
        # Format time string for filename (MMm_SSs)
        minutes = int(current_time_sec // 60)
        seconds = int(current_time_sec % 60)
        time_str = f"{minutes:02d}m_{seconds:02d}s"

        # Save original frame
        filename = os.path.join(input_folder, f"frame_{saved_count:04d}_{time_str}.jpg")
        cv2.imwrite(filename, frame)

        print(f"✅ {time_str} | Frame {saved_count:04d} (Video Frame #{frame_count}) | Saved")

    # Show progress every 10 seconds of video time
    if frame_count % (int(fps) * 10) == 0:   
        print(f"Progress: Reading video... {frame_count/fps:.1f} seconds elapsed")

print("\n Processing Completed Successfully!")
print(f"Total frames saved: {saved_count}")
print(f"Frames saved in: {input_folder}")

cap.release()
```

```python
import cv2
import os
import sys

# ====================== SETTINGS ======================
video_path         = r"C:\Users\Admin\Downloads\projects\vehicle_plate_detection\IN_LANE_3_BACK_ZOOM_NVR_20260429104732_20260429114529_1999865.mp4"
output_folder      = r"C:\Users\Admin\Downloads\projects\vehicle_plate_detection\data\frames3"
max_frames_to_save = 5000
skip_seconds       = 3
# ======================================================

if not os.path.exists(video_path):
    print(f"❌ Video file not found: {video_path}")
    sys.exit()

os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("❌ Could not open video file.")
    sys.exit()

fps          = cap.get(cv2.CAP_PROP_FPS)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

if fps == 0 or total_frames == 0:
    print("❌ Bad FPS/Frame count. Video may be corrupt or unsupported codec.")
    cap.release()
    sys.exit()

print(f"✅ Video Loaded  |  FPS: {fps:.2f}  |  Frames: {total_frames}  |  Duration: {total_frames/fps:.1f}s\n")

frame_count     = 0
saved_count     = 0
last_saved_time = -skip_seconds

while True:
    ret, frame = cap.read()
    if not ret:
        print("End of video reached.")
        break

    frame_count      += 1
    current_time_sec  = frame_count / fps

    if saved_count >= max_frames_to_save:
        print(f"⛔ Limit reached: {max_frames_to_save} frames saved.")
        break

    if current_time_sec >= last_saved_time + skip_seconds:
        last_saved_time = current_time_sec
        saved_count    += 1

        minutes  = int(current_time_sec // 60)
        seconds  = int(current_time_sec % 60)
        time_str = f"{minutes:02d}m_{seconds:02d}s"

        filename = os.path.join(output_folder, f"frame_{saved_count:04d}_{time_str}.jpg")
        cv2.imwrite(filename, frame)
        print(f"💾 {time_str} | Frame {saved_count:04d} (Video Frame #{frame_count})")

    if frame_count % max(1, int(fps) * 10) == 0:
        print(f"⏳ Reading... {current_time_sec:.1f}s elapsed")

cap.release()

print(f"\n✅ Done! {saved_count} frames saved to: {output_folder}")
```
